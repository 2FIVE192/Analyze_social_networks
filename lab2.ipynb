# Лабораторная: центральность по собственным векторам (eigenvector centrality).

Условие: четыре узла, три из которых имеют одинаковую центральность,
а четвёртый — строго большую.

Подобранный граф — звезда K_{1,3}:

        1
        |
    2 - 0 - 3

Узлы 1, 2, 3 симметричны относительно автоморфизмов графа, поэтому их
центральности совпадают; узел 0 — единственный смежный со всеми остальными,
его центральность строго выше.

Аналитически: Ax = lambda x, lambda = sqrt(3),
    центр  = 1/sqrt(2)  ~ 0.707107
    лист   = 1/sqrt(6)  ~ 0.408248
(NetworkX нормирует вектор по евклидовой норме: сумма квадратов = 1.)

## Построение графа

```python
import networkx as nx

G = nx.Graph()
G.add_edges_from([(0, 1), (0, 2), (0, 3)])

print("Узлы:", sorted(G.nodes()))
print("Рёбра:", sorted(G.edges()))
print("Степени:", dict(G.degree()))
```

# Вычисление центральности

```python
# centrality = nx.eigenvector_centrality_numpy(G)

# Вариант на чистом Python (степенной метод), если numpy/scipy недоступны:
centrality = nx.eigenvector_centrality(G, max_iter=1000, tol=1e-10)

print("\nЦентральность по собственным векторам:")
for node, value in sorted(centrality.items(), key=lambda kv: (-kv[1], kv[0])):
    print(f"  узел {node}: {value:.6f}")
```

# Проверка условия

```python
values = sorted(centrality.values())
eps = 1e-9
three_equal = max(values[:3]) - min(values[:3]) < eps
fourth_greater = values[3] - values[2] > eps

print("\nПроверка условия:")
print(f"Три одинаковые ({values[0]:.6f}): {three_equal}")
print(f"Четвёртая больше ({values[3]:.6f}): {fourth_greater}")
print(f"ИТОГ: {'условие выполнено' if three_equal and fourth_greater else 'условие НЕ выполнено'}")
```
